```python
import re
import pandas as pd
from pymatgen.core import Composition

from matminer.featurizers.composition import (
    AtomicOrbitals,
    AtomicPackingEfficiency,
    BandCenter,
    ElementFraction,
    ElementProperty,
    IonProperty,
    Miedema,
    Stoichiometry,
    TMetalFraction,
    ValenceOrbital,
    YangSolidSolution,
)


def _parse_mixture_string(formula_str):
    s = str(formula_str).strip()

    parts = re.findall(r'\(([^()]+)\)\s*([0-9]*\.?[0-9]*)', s)

    if parts:
        return [
            (formula.strip(), float(frac) if frac else 1.0)
            for formula, frac in parts
        ]

    return [(s, 1.0)]


def _weighted_mixture_composition(formula_str):
    mix = Composition()
    for formula, frac in _parse_mixture_string(formula_str):
        mix += Composition(formula) * frac
    return mix


def featurize_pretty_formula_series(pretty_formula_series, ignore_errors=True):
    """
    Returns a DataFrame with a unique row_id preserved so the user can
    safely merge results back even when formulas repeat.
    """
    # ensure we keep unique identity
    df = pd.DataFrame({
        "row_id": pretty_formula_series.index,
        "pretty_formula": pretty_formula_series.values
    })

    df["composition"] = df["pretty_formula"].apply(_weighted_mixture_composition)

    featurizers = [
        AtomicOrbitals(),
        AtomicPackingEfficiency(),
        BandCenter(),
        ElementFraction(),
        ElementProperty.from_preset("magpie"),
        IonProperty(),
        Miedema(),
        Stoichiometry(),
        TMetalFraction(),
        ValenceOrbital(),
        YangSolidSolution(),
    ]

    for feat in featurizers:
        df = feat.featurize_dataframe(
            df,
            "composition",
            ignore_errors=ignore_errors
        )

    return df
```

```python

def drop_most_correlated_features(
    X: pd.DataFrame,
    threshold: float = 0.85,
    method: str = "pearson",
):
    """
    Iteratively drops the feature with the largest number of
    pairwise correlations above the threshold.

    This only uses X, not the targets Y.
    """

    X_work = X.copy()
    dropped_features = []
    history = []

    while True:
        corr = X_work.corr(method=method).abs()

        # Ignore self-correlation
        np.fill_diagonal(corr.values, 0)

        # Count how many correlations each feature has above threshold
        high_corr_counts = (corr > threshold).sum(axis=0)

        max_count = high_corr_counts.max()

        # Stop when no pairwise correlation is above threshold
        if max_count == 0:
            break

        feature_to_drop = high_corr_counts.idxmax()

        history.append({
            "dropped_feature": feature_to_drop,
            "number_of_correlations_above_threshold": int(max_count),
            "correlated_with": corr.index[corr[feature_to_drop] > threshold].tolist(),
        })

        X_work = X_work.drop(columns=[feature_to_drop])
        dropped_features.append(feature_to_drop)

    return X_work, dropped_features, history



def split_XY(df):
    """
    Splits a dataframe into X (features) and Y (targets).

    Y columns:
        - Electrical Conductivity (S/cm)
        - Seebeck Coefficient (µV/K)
        - zT
        - Total Thermal Conductivity (W/mK)
    X contains all other columns.
    """

    y_cols = [
    "Electrical Conductivity (S/cm)",
    "Seebeck Coefficient (µV/K)",
    "zT",
    "Total Thermal Conductivity (W/mK)"]


    Y = df[y_cols].copy()
    X = df.drop(columns=y_cols).copy()

    return X, Y






df_clean=pd.read_csv(r'CLEANED_DATA.csv')
print(df_clean.head())
df_clean = df_clean.loc[:, ~df_clean.columns.str.contains("^Unnamed")]
print(df_clean.head())


X_all, Y_all = split_XY(df_clean)


X_all = X_all.copy()
Y_all = Y_all.copy()

X_all.index = df_clean.index
Y_all.index = df_clean.index


# ---------------------------------------------------------
# 3. Apply correlation filtering only to X
# ---------------------------------------------------------

X_filtered, dropped_features, corr_history = drop_most_correlated_features(
    X_all,
    threshold=0.85,
    method="pearson",
)


# ---------------------------------------------------------
# 4. Stack filtered X and original Y back together
# ---------------------------------------------------------

df_clean_filtered = pd.concat([X_filtered, Y_all], axis=1)


print("Original df_clean shape:", df_clean.shape)
print("Original X shape:", X_all.shape)
print("Filtered X shape:", X_filtered.shape)
print("Filtered full dataframe shape:", df_clean_filtered.shape)

print("Number of dropped features:", len(dropped_features))
print("Dropped features:")
print(dropped_features)

corr_history_df = pd.DataFrame(corr_history)
print(corr_history_df.head())
```
